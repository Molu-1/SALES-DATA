## Advanced Options

Beyond local editing two things are quite important: editing the notebooks in the markdown format and running Jupyter remotely.
The latter matters when we want to run the code on a faster server.
The former matters since Jupyter's native ipynb format stores a lot of auxiliary data that is
irrelevant to the content,
mostly related to how and where the code is run.
This is confusing for Git, making
reviewing contributions very difficult.
Fortunately there is an alternative---native editing in the markdown format.

### Markdown Files in Jupyter

If you wish to contribute to the content of this book, you need to modify the
source file (md file, not ipynb file) on GitHub.
Using the notedown plugin we
can modify notebooks in the md format directly in Jupyter.


First, install the notedown plugin, run the Jupyter Notebook, and load the plugin:

```
pip install d2l-notedown  # You may need to uninstall the original notedown.
jupyter notebook --NotebookApp.contents_manager_class='notedown.NotedownContentsManager'
```

You may also turn on the notedown plugin by default whenever you run the Jupyter Notebook.
First, generate a Jupyter Notebook configuration file (if it has already been generated, you can skip this step).

```
jupyter notebook --generate-config
```

Then, add the following line to the end of the Jupyter Notebook configuration file (for Linux or macOS, usually in the path `~/.jupyter/jupyter_notebook_config.py`):

```
c.NotebookApp.contents_manager_class = 'notedown.NotedownContentsManager'
```

After that, you only need to run the `jupyter notebook` command to turn on the notedown plugin by default.

### Running Jupyter Notebooks on a Remote Server

Sometimes, you may want to run Jupyter notebooks on a remote server and access it through a browser on your local computer. If Linux or macOS is installed on your local machine (Windows can also support this function through third-party software such as PuTTY), you can use port forwarding:

```
ssh myserver -L 8888:localhost:8888
```

The above string `myserver` is the address of the remote server.
Then we can use http://localhost:8888 to access the remote server `myserver` that runs Jupyter notebooks. We will detail on how to run Jupyter notebooks on AWS instances
later in this appendix.

### Timing

We can use the `ExecuteTime` plugin to time the execution of each code cell in Jupyter notebooks.
Use the following commands to install the plugin:

```
pip install jupyter_contrib_nbextensions
jupyter contrib nbextension install --user
jupyter nbextension enable execute_time/ExecuteTime
```

````python
import os, zipfile, glob
import pandas as pd
import numpy as np
from google.colab import files

uploaded = files.upload()  # Upload archive (4)(1).zip OR the .xlsx file

excel_path = None
for filename in uploaded:
    if filename.lower().endswith(".zip"):
        with zipfile.ZipFile(filename, "r") as z:
            excel_files = [n for n in z.namelist() if n.lower().endswith((".xlsx", ".xls"))]
            if not excel_files:
                raise ValueError("No Excel file found inside the ZIP.")
            z.extract(excel_files[0], "/content")
            excel_path = os.path.join("/content", excel_files[0])
    elif filename.lower().endswith((".xlsx", ".xls")):
        excel_path = filename

if excel_path is None:
    raise ValueError("Please upload a ZIP or Excel file.")

df_raw = pd.read_excel(excel_path)
df = df_raw.copy()
print("Dataset loaded successfully!")
print("Rows:", len(df), "| Columns:", len(df.columns))
display(df.head())
````

````python
## Step 4 — Data cleaning and quality checks
````

````python
# Standardize column names (keep readable snake_case)
df.columns = [str(c).strip().lower().replace(" ", "_") for c in df.columns]
````

````python
# Remove exact duplicate rows
before = len(df)
df = df.drop_duplicates().copy()
print("Exact duplicate rows removed:", before - len(df))
````

````python
# Clean text columns
text_cols = df.select_dtypes(include=["object"]).columns
for col in text_cols:
    df[col] = df[col].astype("string").str.strip()
    df[col] = df[col].replace({"": pd.NA, "nan": pd.NA, "None": pd.NA})
````

````python
# Convert likely numeric fields safely
numeric_candidates = [
    "age", "daily_screen_time_hours", "social_media_hours",
    "gaming_hours", "sleep_hours"
]
for col in numeric_candidates:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")
````

````python
# Missing values report
quality_report = pd.DataFrame({
    "Missing values": df.isna().sum(),
    "Missing %": (df.isna().mean() * 100).round(2),
    "Unique values": df.nunique(dropna=True)
}).sort_values("Missing %", ascending=False)

display(quality_report)
print("Duplicate user IDs:", int(df["user_id"].duplicated(keep=False).sum()) if "user_id" in df.columns else "user_id column not found")
````

````python
## Step 5 — Create analysis metrics
````

````python
def get_mean(col):
    return df[col].mean() if col in df.columns else np.nan

total_records = len(df)
unique_users = df["user_id"].nunique() if "user_id" in df.columns else np.nan
avg_screen_time = get_mean("daily_screen_time_hours")
avg_sleep = get_mean("sleep_hours")
avg_social = get_mean("social_media_hours")
avg_gaming = get_mean("gaming_hours")

print(f"📌 Total records: {total_records:,}")
print(f"👥 Unique users: {unique_users:,.0f}" if pd.notna(unique_users) else "Unique users: N/A")
print(f"📱 Average daily screen time: {avg_screen_time:.2f} hours")
print(f"😴 Average sleep: {avg_sleep:.2f} hours")
print(f"💬 Average social media time: {avg_social:.2f} hours")
print(f"🎮 Average gaming time: {avg_gaming:.2f} hours")
````

````python
import plotly.express as px
import plotly.graph_objects as go
import ipywidgets as widgets
from IPython.display import display, clear_output

# Build filter options from the actual dataset
category_col = "screen_time_category" if "screen_time_category" in df.columns else None
gender_col = "gender" if "gender" in df.columns else None
stress_col = "stress_level" if "stress_level" in df.columns else None

gender_filter = widgets.SelectMultiple(
    options=sorted(df[gender_col].dropna().astype(str).unique()) if gender_col else [],
    value=tuple(sorted(df[gender_col].dropna().astype(str).unique())) if gender_col else (),
    description="Gender", layout=widgets.Layout(width="32%", height="90px")
)
stress_filter = widgets.SelectMultiple(
    options=sorted(df[stress_col].dropna().astype(str).unique()) if stress_col else [],
    value=tuple(sorted(df[stress_col].dropna().astype(str).unique())) if stress_col else (),
    description="Stress", layout=widgets.Layout(width="32%", height="90px")
)
age_slider = widgets.IntRangeSlider(
    value=(int(df.age.min()), int(df.age.max())) if "age" in df else (0,100),
    min=int(df.age.min()) if "age" in df else 0,
    max=int(df.age.max()) if "age" in df else 100,
    description="Age", continuous_update=False, layout=widgets.Layout(width="95%")
)
out = widgets.Output()

def render_dashboard(*_):
    with out:
        clear_output(wait=True)
        filtered = df.copy()
        if gender_col and gender_filter.value:
            filtered = filtered[filtered[gender_col].astype(str).isin(gender_filter.value)]
        if stress_col and stress_filter.value:
            filtered = filtered[filtered[stress_col].astype(str).isin(stress_filter.value)]
        if "age" in filtered:
            filtered = filtered[filtered.age.between(*age_slider.value)]

        if filtered.empty:
            print("No records match these filters.")
            return

        st = filtered["daily_screen_time_hours"].mean() if "daily_screen_time_hours" in filtered else np.nan
        sl = filtered["sleep_hours"].mean() if "sleep_hours" in filtered else np.nan
        display(HTML(f'<h3>Filtered records: {len(filtered):,} &nbsp; | &nbsp; Avg screen time: {st:.2f} hrs &nbsp; | &nbsp; Avg sleep: {sl:.2f} hrs</h3>'))

        if category_col:
            fig = px.histogram(filtered, x=category_col, color=category_col,
                               title="Users by Screen-Time Category", text_auto=True)
            fig.update_layout(template="plotly_white", showlegend=False)
            fig.show()

        if stress_col and "daily_screen_time_hours" in filtered:
            fig = px.box(filtered, x=stress_col, y="daily_screen_time_hours", color=stress_col,
                         title="Screen Time Distribution by Stress Level")
            fig.update_layout(template="plotly_white", showlegend=False)
            fig.show()

        if "daily_screen_time_hours" in filtered and "sleep_hours" in filtered:
            fig = px.scatter(filtered, x="daily_screen_time_hours", y="sleep_hours",
                             color=stress_col if stress_col else None,
                             hover_data=[c for c in ["user_id","age","gender"] if c in filtered],
                             trendline="ols", opacity=.65,
                             title="Daily Screen Time vs Sleep Hours")
            fig.update_layout(template="plotly_white")
            fig.show()

        numeric = filtered.select_dtypes(include=np.number)
        if numeric.shape[1] >= 2:
            corr = numeric.corr(numeric_only=True)
            fig = px.imshow(corr, text_auto=".2f", color_continuous_scale="RdBu_r",
                            zmin=-1, zmax=1, title="Correlation Heatmap")
            fig.update_layout(template="plotly_white")
            fig.show()

        if "age" in filtered and "daily_screen_time_hours" in filtered:
            age_summary = filtered.groupby(pd.cut(filtered.age, bins=[0,18,25,35,45,55,120],
                                                  labels=["Under 18","18–25","26–35","36–45","46–55","56+"]),
                                           observed=False)["daily_screen_time_hours"].mean().reset_index()
            age_summary.columns = ["Age Group","Average Screen Time"]
            fig = px.bar(age_summary, x="Age Group", y="Average Screen Time",
                         title="Average Screen Time by Age Group", text_auto=".2f")
            fig.update_layout(template="plotly_white")
            fig.show()

gender_filter.observe(render_dashboard, names="value")
stress_filter.observe(render_dashboard, names="value")
age_slider.observe(render_dashboard, names="value")
display(widgets.HBox([gender_filter, stress_filter]))
display(age_slider, out)
render_dashboard()
````

````python
# Midnight / dark visual theme
DARK = 'plotly_dark'
ACCENT = ['#38bdf8','#a78bfa','#34d399','#fbbf24','#fb7185']
px.defaults.template = DARK

filter_columns = ['gender','Age_group','Screen_time_category','stress_level','Sleep_Health']
filter_controls = {}
for col in filter_columns:
    if col in df.columns:
        vals = sorted(df[col].dropna().astype(str).unique().tolist())
        filter_controls[col] = widgets.Dropdown(options=['All']+vals, value='All',
            description=col.replace('_',' ').title(),
            style={'description_width':'initial'},
            layout=widgets.Layout(width='31%', margin='4px'))

refresh = widgets.Button(description='Refresh dashboard', icon='refresh', button_style='info',
                         layout=widgets.Layout(width='210px', height='42px'))
output = widgets.Output()

def selected_data():
    d = df.copy()
    for col, control in filter_controls.items():
        if control.value != 'All': d = d[d[col].astype(str) == control.value]
    return d

def metric_card(label, value, note):
    return f"""<div style="background:linear-gradient(145deg,#172554,#111827);border:1px solid #334155;
    border-radius:16px;padding:18px 20px;min-width:175px;flex:1;box-shadow:0 8px 25px #02061755">
    <div style="color:#94a3b8;font-size:11px;letter-spacing:1px;text-transform:uppercase;font-weight:700">{label}</div>
    <div style="color:#f8fafc;font-size:29px;font-weight:800;margin:9px 0 4px">{value}</div>
    <div style="color:#64748b;font-size:11px">{note}</div></div>"""

def show_dashboard(_=None):
    with output:
        clear_output(wait=True)
        d = selected_data()
        if d.empty:
            display(HTML('<h3 style="color:#f87171">No data matches the selected filters.</h3>')); return
        n_users = d['user_id'].nunique() if 'user_id' in d else len(d)
        avg_screen = d['daily_screen_time_hours'].mean() if 'daily_screen_time_hours' in d else np.nan
        avg_sleep = d['sleep_hours'].mean() if 'sleep_hours' in d else np.nan
        poor_sleep = (d['Sleep_Health'].astype(str).str.contains('poor', case=False, na=False).mean()*100) if 'Sleep_Health' in d else np.nan
        cards = [
            metric_card('Records',f'{len(d):,}','Filtered dataset rows'),
            metric_card('Unique users',f'{n_users:,}','Distinct user IDs'),
            metric_card('Average screen time',f'{avg_screen:.2f} h' if pd.notna(avg_screen) else 'N/A','Hours per day'),
            metric_card('Average sleep',f'{avg_sleep:.2f} h' if pd.notna(avg_sleep) else 'N/A','Hours per day'),
            metric_card('Poor sleep share',f'{poor_sleep:.1f}%' if pd.notna(poor_sleep) else 'N/A','Based on Sleep_Health label')]
        display(HTML('<div style="display:flex;flex-wrap:wrap;gap:12px;margin:10px 0 22px">'+''.join(cards)+'</div>'))

        # Chart 1: screen-time category mix
        if 'Screen_time_category' in d:
            s = d['Screen_time_category'].value_counts(dropna=False).rename_axis('Category').reset_index(name='Users')
            fig = px.bar(s, x='Category', y='Users', color='Category', title='01 · Screen-time category mix', text='Users', color_discrete_sequence=ACCENT)
            fig.update_layout(showlegend=False); fig.show()

        # Chart 2: sleep health split
        if 'Sleep_Health' in d:
            s = d['Sleep_Health'].value_counts(dropna=False).rename_axis('Sleep health').reset_index(name='Records')
            fig = px.pie(s, names='Sleep health', values='Records', hole=.62, title='02 · Sleep health composition', color_discrete_sequence=ACCENT)
            fig.update_traces(textposition='inside', textinfo='percent+label'); fig.show()

        # Chart 3: daily screen vs sleep with trendline
        if {'daily_screen_time_hours','sleep_hours'}.issubset(d.columns):
            fig = px.scatter(d, x='daily_screen_time_hours', y='sleep_hours', color='stress_level' if 'stress_level' in d else None,
                hover_data=[c for c in ['user_id','age','gender'] if c in d], opacity=.72,
                title='03 · Screen time and sleep relationship', trendline='ols' if len(d)>2 else None,
                color_discrete_sequence=ACCENT)
            fig.update_layout(xaxis_title='Screen time (hours/day)', yaxis_title='Sleep (hours/day)'); fig.show()

        # Chart 4: activity time mix
        activity_cols = [c for c in ['social_media_hours','gaming_hours'] if c in d]
        if activity_cols:
            act = d[activity_cols].mean().rename_axis('Activity').reset_index(name='Average hours')
            act['Activity'] = act['Activity'].map({'social_media_hours':'Social media','gaming_hours':'Gaming'})
            fig = px.bar(act, x='Activity', y='Average hours', color='Activity', text_auto='.2f', title='04 · Average daily activity hours', color_discrete_sequence=ACCENT)
            fig.update_layout(showlegend=False); fig.show()

        # Chart 5: stress profile by age group
        if {'Age_group','stress_level'}.issubset(d.columns):
            cross = pd.crosstab(d['Age_group'], d['stress_level'], normalize='index')*100
            fig = px.bar(cross.reset_index().melt(id_vars='Age_group', var_name='Stress level', value_name='Share %'),
                x='Age_group', y='Share %', color='Stress level', barmode='stack', title='05 · Stress profile across age groups', color_discrete_sequence=ACCENT)
            fig.update_layout(yaxis_title='Share of records (%)'); fig.show()

        # Chart 6: average screen time by gender and age
        if {'gender','Age_group','daily_screen_time_hours'}.issubset(d.columns):
            g = d.groupby(['Age_group','gender'], observed=False)['daily_screen_time_hours'].mean().reset_index()
            fig = px.density_heatmap(g, x='gender', y='Age_group', z='daily_screen_time_hours', histfunc='avg',
                title='06 · Average screen time by age group and gender', color_continuous_scale='Teal')
            fig.show()

        # Compact insights table
        display(HTML('<h3 style="color:#e2e8f0">Quick insight table</h3>'))
        numeric = [c for c in ['daily_screen_time_hours','sleep_hours','social_media_hours','gaming_hours'] if c in d]
        if numeric: display(d[numeric].describe().round(2))
        if {'daily_screen_time_hours','sleep_hours'}.issubset(d.columns):
            corr = d['daily_screen_time_hours'].corr(d['sleep_hours'])
            display(HTML(f'<p style="color:#cbd5e1"><b>Screen-time / sleep correlation:</b> {corr:.3f}. This is an association, not proof of cause.</p>'))

refresh.on_click(show_dashboard)
filter_ui = widgets.HBox(list(filter_controls.values()), layout=widgets.Layout(flex_flow='row wrap'))
display(widgets.VBox([filter_ui, refresh, output]))
show_dashboard()
````
